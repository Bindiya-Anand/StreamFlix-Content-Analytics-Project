# Phase 1: Data Cleaning & Quality Check

## Loading Datasets

```python
import pandas as pd

# Define files and names
file_paths = {
    "subscribers": "/content/subscribers.csv",
    "titles": "/content/titles.csv",
    "watch_history": "/content/watch_history.csv",
    "ratings": "/content/ratings.csv",
    "reviews": "/content/reviews.csv",
    "watchlist": "/content/watchlist.csv"
}

# Load all datasets into the datasets dictionary
datasets = {}
for name, path in file_paths.items():
    try:
        datasets[name] = pd.read_csv(path)
        print(f"Successfully loaded {name} ({datasets[name].shape[0]} rows, {datasets[name].shape[1]} columns)")
    except Exception as e:
        print(f"Error loading {name}: {e}")

# Assign back to individual global variables for compatibility
subscribers = datasets.get("subscribers")
titles = datasets.get("titles")
watch_history = datasets.get("watch_history")
ratings = datasets.get("ratings")
reviews = datasets.get("reviews")
watchlist = datasets.get("watchlist")

# Display the head of each loaded dataset to verify
for name, df in datasets.items():
    if df is not None:
        print(f"\n=== Head of {name} ===")
        display(df.head())
        print("="*50)
```

## Dataset Dimension Summary

```python
shape_records = []
for name, df in datasets.items():
    shape_records.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    })

shape_summary_df = pd.DataFrame(shape_records)
display(shape_summary_df)
```

## Check dtypes of columns

```python
# Print dtypes of all columns for each dataset to verify current states
for dataset_name, df in datasets.items():
    print(f"=== Column Data Types in '{dataset_name}' ===")
    display(df.dtypes.to_frame(name='Data Type'))
    print("\n" + "="*40 + "\n")
```

## Converting Date attributes to datetime dtype

```python
# Map each dataset to its corresponding date columns that need conversion
date_columns_mapping = {
    'subscribers': ['signup_date', 'churn_date'],
    'titles': ['date_added', 'license_expiry'],
    'watch_history': ['watch_date'],
    'ratings': ['rating_date'],
    'reviews': ['review_date'],
    'watchlist': ['added_date']
}

# Perform permanent conversions in-place directly on the DataFrames inside our active datasets dictionary
for dataset_name, columns in date_columns_mapping.items():
    df = datasets[dataset_name]
    print(f"--- Converting date columns in {dataset_name} ---")
    for col in columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')
            print(f"Converted '{col}' to {df[col].dtype}")
    print()

# Synchronize global variables to point to these permanently converted DataFrames
subscribers = datasets.get("subscribers")
titles = datasets.get("titles")
watch_history = datasets.get("watch_history")
ratings = datasets.get("ratings")
reviews = datasets.get("reviews")
watchlist = datasets.get("watchlist")
```

## Checking for Null Values

```python
datasets = {
    "subscribers": subscribers,
    "titles": titles,
    "watch_history": watch_history,
    "ratings": ratings,
    "reviews": reviews,
    "watchlist": watchlist
}

for name, df in datasets.items():
    print(f"=== Null Values in {name} ===")
    total_rows = len(df)
    null_counts = df.isna().sum()
    columns_with_nulls = null_counts[null_counts > 0]

    if len(columns_with_nulls) > 0:
        # Create a summary DataFrame with both count and percentage
        summary_df = pd.DataFrame({
            'Missing Count': columns_with_nulls,
            'Null %': (columns_with_nulls / total_rows * 100).round(2)
        })
        display(summary_df)
        print()
    else:
        print("No missing values found!\n")
```

## Outlier Detection - `watch_duration_min `

```python
# 1. Identify sessions where watch duration exceeds the actual content duration
longer_than_content = watch_history[watch_history['watch_duration_min'] > watch_history['content_duration_min']]

count_flagged = len(longer_than_content)
pct_flagged = (count_flagged / len(watch_history)) * 100

print(f"Number of sessions longer than content duration: {count_flagged}")
print(f"Percentage of flagged sessions: {pct_flagged:.4f}%")

# 2. Display a sample of these flagged sessions if they exist
if count_flagged > 0:
    print("\n--- Sample of Flagged Sessions ---")
    display(longer_than_content.head())
else:
    print("\nNo sessions were found where the watch duration exceeded the content duration.")
```

## Referential Integrity - `subscriber_id` and `title_id` in `watch_history.csv`

```python
# 1. Check if all subscriber_id in watch_history exist in subscribers
missing_subscribers = watch_history[~watch_history['subscriber_id'].isin(subscribers['subscriber_id'])]['subscriber_id'].unique()

# 2. Check if all title_id in watch_history exist in titles
missing_titles = watch_history[~watch_history['title_id'].isin(titles['title_id'])]['title_id'].unique()

print(f"Number of orphaned subscriber_id in watch_history: {len(missing_subscribers)}")
if len(missing_subscribers) > 0:
    print("Orphaned subscriber IDs:", missing_subscribers)

print(f"\nNumber of orphaned title_id in watch_history: {len(missing_titles)}")
if len(missing_titles) > 0:
    print("Orphaned title IDs:", missing_titles)
```

## Specific Checks

### Active Subscriber Integrity Check: `is_active` vs `churn_date` in Subscribers

```python
# Identify active subscribers who incorrectly have a non-null churn_date
active_with_churn_date = subscribers[
    (subscribers['is_active'] == True) &
    (subscribers['churn_date'].notna())
]

count_active_mismatches = len(active_with_churn_date)
print(f"Number of active subscribers with a churn date: {count_active_mismatches}")

if count_active_mismatches > 0:
    print("\n--- Flagged Active Subscribers with Churn Dates ---")
    display(active_with_churn_date[['subscriber_id', 'signup_date', 'churn_date', 'is_active']])
else:
    print("\nValidation Passed: All active subscribers have a blank churn_date.")
```

### Date Consistency Check: `churn_date` vs `signup_date` in Subscribers

```python
# Identify subscribers where churn_date is on or before signup_date
invalid_churn_dates = subscribers[
    (subscribers['churn_date'].notna()) &
    (subscribers['churn_date'] <= subscribers['signup_date'])
]

count_invalid = len(invalid_churn_dates)
print(f"Number of subscribers with churn_date on or before signup_date: {count_invalid}")

if count_invalid > 0:
    print("\n--- Flagged Rows with Invalid Churn Dates ---")
    display(invalid_churn_dates[['subscriber_id', 'signup_date', 'churn_date', 'is_active']])
else:
    print("\nValidation Passed: All churned subscribers have a churn_date after their signup_date.")
```

### Logical Consistency Check: `completion_pct` Calculation in Watch History

```python
# Calculate the expected completion percentage based on watch and content durations
expected_completion = (watch_history['watch_duration_min'] / watch_history['content_duration_min']) * 100

# Measure the absolute difference between the reported and calculated percentages
diff = (watch_history['completion_pct'] - expected_completion).abs()

# Define a tolerance threshold (e.g., 1% due to potential rounding differences)
tolerance = 1.0
mismatches = watch_history[diff > tolerance]

count_mismatches = len(mismatches)
pct_mismatches = (count_mismatches / len(watch_history)) * 100

print(f"Number of mismatched records (difference > {tolerance}%): {count_mismatches} ({pct_mismatches:.4f}% of total)")

if count_mismatches > 0:
    print("\n--- Sample of Flagged Mismatches ---")
    temp_df = mismatches.copy()
    temp_df['Expected_Pct'] = expected_completion[diff > tolerance]
    display(temp_df[['watch_id', 'watch_duration_min', 'content_duration_min', 'completion_pct', 'Expected_Pct']].head())
else:
    print("\nValidation Passed: All watch sessions have completion percentages matching their duration ratios.")
```

### Sentiment Analysis Validation: `sentiment` Column Integrity in Reviews

```python
# Define the allowed set of sentiments
allowed_sentiments = {'Positive', 'Neutral', 'Negative'}

# Identify any rows where the sentiment does not match the allowed list
invalid_sentiment_mask = ~reviews['sentiment'].isin(allowed_sentiments)
invalid_sentiments = reviews[invalid_sentiment_mask]

count_invalid = len(invalid_sentiments)
print(f"Number of reviews with invalid sentiment values: {count_invalid}")

if count_invalid > 0:
    print("\n--- Unique Invalid Sentiments Found ---")
    print(invalid_sentiments['sentiment'].unique())
    print("\n--- Sample of Flagged Rows ---")
    display(invalid_sentiments.head())
else:
    print("\nValidation Passed: All sentiments are strictly either 'Positive', 'Neutral', or 'Negative'.")
```

### Referential Integrity Check: `watchlist.csv` link to `subscribers` and `titles`

```python
# 1. Verify all subscriber_id in watchlist exist in subscribers
missing_watchlist_subs = watchlist[~watchlist['subscriber_id'].isin(subscribers['subscriber_id'])]['subscriber_id'].unique()

# 2. Verify all title_id in watchlist exist in titles
missing_watchlist_titles = watchlist[~watchlist['title_id'].isin(titles['title_id'])]['title_id'].unique()

print(f"Orphaned subscriber_id records in watchlist: {len(missing_watchlist_subs)}")
if len(missing_watchlist_subs) > 0:
    print("Sample orphaned subscriber IDs:", missing_watchlist_subs[:10])

print(f"Orphaned title_id records in watchlist: {len(missing_watchlist_titles)}")
if len(missing_watchlist_titles) > 0:
    print("Sample orphaned title IDs:", missing_watchlist_titles[:10])

if len(missing_watchlist_subs) == 0 and len(missing_watchlist_titles) == 0:
    print("\nValidation Passed: Every watchlist entry successfully links to an existing subscriber and title.")
```

### Tenure Check: Active Subscribers with Very Short Tenure

```python
# Identify active subscribers (is_active == True) with 0 or 1 month of tenure
short_tenure_active = subscribers[(subscribers['is_active'] == True) & (subscribers['tenure_months'] <= 1)]

count_short_active = len(short_tenure_active)
pct_short_active = (count_short_active / len(subscribers)) * 100

print(f"Number of active subscribers with tenure <= 1 month: {count_short_active} ({pct_short_active:.2f}% of total subscribers)")

# Display a sample of these active new subscribers
if count_short_active > 0:
    display(short_tenure_active[['subscriber_id', 'signup_date', 'tenure_months', 'is_active', 'plan_type']].head())
else:
    print("No active subscribers found with tenure <= 1 month.")
```

### Investigating Null Values in `license_expiry`

```python
# Flag missing license expiry dates
titles['expiry_missing'] = titles['license_expiry'].isna()

# Check if missingness is related to the title 'type' (Movie vs. TV Show)
type_vs_expiry = pd.crosstab(
    titles['type'],
    titles['expiry_missing'],
    rownames=['Type'],
    colnames=['License Expiry is Null']
)

print("=== Missing License Expiry by Title Type ===")
display(type_vs_expiry)

# Let's also check if it varies significantly by 'primary_genre'
genre_vs_expiry = pd.crosstab(
    titles['primary_genre'],
    titles['expiry_missing'],
    rownames=['Primary Genre'],
    colnames=['License Expiry is Null']
)

print("\n=== Missing License Expiry by Primary Genre ===")
display(genre_vs_expiry)
```
